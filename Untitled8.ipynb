```python
import json
import os

json_path = os.path.join(
    new_path,
    "annotations_train.json"
)

with open(json_path, "r") as f:
    data = json.load(f)

print("Total Images:", len(data))
```

```python
states = set()

for value in data.values():
    states.add(value["driver_state"])

print(states)
```

```python
import pandas as pd

rows = []

for img_path, info in data.items():

    full_path = os.path.join(
        new_path,
        img_path.replace("./classification_frames/","")
    )

    if info["driver_state"] == "alert":
        label = "Not_Drowsy"
    else:
        label = "Drowsy"

    rows.append([
        full_path,
        label
    ])

df = pd.DataFrame(
    rows,
    columns=[
        "image_path",
        "label"
    ]
)

df.head()
```

```python
df.to_csv(
    "drowsiness_dataset.csv",
    index=False
)

print("CSV Saved")
```

```python
print(df["label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df["label"],
    random_state=42
)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_gen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

test_gen = ImageDataGenerator(
    rescale=1./255
)
```

```python
train_data = train_gen.flow_from_dataframe(
    train_df,
    x_col="image_path",
    y_col="label",
    target_size=(128,128),
    batch_size=32,
    class_mode="binary"
)

test_data = test_gen.flow_from_dataframe(
    test_df,
    x_col="image_path",
    y_col="label",
    target_size=(128,128),
    batch_size=32,
    class_mode="binary"
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import *
```

```python
model = Sequential()

model.add(
    Conv2D(
        32,
        (3,3),
        activation='relu',
        input_shape=(128,128,3)
    )
)

model.add(MaxPooling2D())

model.add(
    Conv2D(
        64,
        (3,3),
        activation='relu'
    )
)

model.add(MaxPooling2D())

model.add(
    Conv2D(
        128,
        (3,3),
        activation='relu'
    )
)

model.add(MaxPooling2D())

model.add(
    Conv2D(
        256,
        (3,3),
        activation='relu'
    )
)

model.add(MaxPooling2D())

model.add(
    Reshape((36,256))
)

model.add(
    LSTM(128)
)

model.add(
    Dense(
        128,
        activation='relu'
    )
)

model.add(
    Dropout(0.5)
)

model.add(
    Dense(
        1,
        activation='sigmoid'
    )
)
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=5
)
```
