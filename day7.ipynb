```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
vocab_size=10000
(X_train,y_train),(X_test,y_test)=imdb.load_data(num_words=vocab_size)
print("Training samples",len(X_train))
print("Testing samples",len(X_test))
```

```python
max_length=200
X_train=pad_sequences(X_train,maxlen=max_length,padding='post')
X_test=pad_sequences(X_test,maxlen=max_length,padding='post')
```

```python
model=Sequential([
    Embedding(input_dim=vocab_size,output_dim=32,input_length=max_length),
    SimpleRNN(64),
    Dense(32,activation='relu'),
    Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
history=model.fit(X_train,y_train,validation_split=0.2,epochs=5,batch_size=64)
```

```python
loss,accuracy=model.evaluate(
    X_test,y_test
)
print("Accuracy:",accuracy)
```

```python
from tensorflow.keras.datasets import imdb
word_index=imdb.get_word_index()
word_index={k: (v+3) for k,v in word_index.items()}
word_index["<PAD>"]=0
word_index["<START>"]=1
word_index["<UNK>"]=2
word_index["<UNUSED>"]=3
```

```python
def review_to_sequence(review):
  review=review.lower().split()
  sequence=[]
  for word in review:
    sequence.append(word_index.get(word,2))
  return sequence
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
max_length=200
def predict_sentiment(review):
  sequence=review_to_sequence(review)
  padded=pad_sequences([sequence],maxlen=max_length,padding='post')
  prediction=model.predict(padded,verbose=0)
  score=prediction[0][0]
  print("Sentiment Score:",score)
  if score >0.5:
    print("Positive Review")
  else:
    print("Negative Review")
```

```python
predict_sentiment("The service was excellent.")
```
