```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoded_data=df.copy()
```

```python
label_encoder=LabelEncoder()
encoded_data['Gender_Label']=label_encoder.fit_transform(encoded_data['Gender'])
```

```python
one_hot_encoded=pd.get_dummies(encoded_data['Department'],prefix='Department')
encoded_data=pd.concat([encoded_data,one_hot_encoded],axis=1)
```

```python
Experience_Years_mapping={1:'First',2:'Second',3:'Third'}
encoded_data['Experience_Years_Category']=encoded_data['Experience_Years'].map(Experience_Years_mapping)
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
numerical_columns=['Employee_ID','Age','Experience_Years','Monthly_Salary']
```

```python
df.dropna(subset=numerical_columns,inplace=True)
```

```python
minmax_scaler=MinMaxScaler()
standard_scaler=StandardScaler()
```

```python
df['Age_MinMax']=minmax_scaler.fit_transform(df[['Age']])
df['Monthly_Salary_MinMax']=minmax_scaler.fit_transform(df[['Monthly_Salary']])
```

```python
df['Age_Standard']=standard_scaler.fit_transform(df[['Age']])
df['Monthly_Salary_Standard']=standard_scaler.fit_transform(df[['Monthly_Salary']])
```
